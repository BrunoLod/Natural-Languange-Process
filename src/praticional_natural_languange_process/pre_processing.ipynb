### Preparando o texto 

A preparação de um texto para ser utilizado para contextos de NLP podem ser compreendidos a partir das seguintes etapas principais : 

- definir formato das letras - normalmente em lower case
- remoção de pontuações e stopwords (termos relativos artigos, preposição e afins)
- normalização - com uso de stemming ou lemmatization

**Pontos de atenção:**

Stemming e lemmatization se refere às técnicas utilizadas para normalizar os textos, de modo a tornar os termos presentes no texto segundo o seu correspondente elementar, capaz de conter a característifica fundamental que simboliza o significado daquele significante. Por significante entenda o nome do termo enquanto tal, sendo o significado o sentido que aquele nome representa ou se refere. Por exemplo, carro significa um tipo de veículo capaz de se mover de forma autônoma ou com ingerência das pessoas, enquanto o seu nome carro, abstendo-se de seu significado representa o significante. 

No entanto, por mais que ambas as técnicas objetivam realizar semelhante fim, apresentam certas difenças. O stemming mapeia o termo qualquer para a sua unidade mais radical, enquanto que o lemmatization pode mapeá-lo para outro que realiza essa função. À título de exemplos, pode-se compreender stemming e lemmatization da seguinte forma : 

- stemming : caminhando - caminhar
- lemmatization: caminhando - andar



````python
# Passando o texto para lower case.

sentence_lower = sentence.lower()

print(sentence_lower)
````

````python
# Removendo as pontuações.

corpus = sentence_lower.translate(str.maketrans("", "", punctuation))

print(corpus)
````

````python
# Removendo stop words.

stop_words = set(stopwords.words("english"))

tokenized_corpus = word_tokenize(corpus)
print(f"Corpus com stopwords: \n{tokenized_corpus}\n\n")

tokenized_corpus_without_stopwords = [i for i in tokenized_corpus if not i in stop_words]
print(f"Corpus sem stopwords: \n{tokenized_corpus_without_stopwords}")

````

Como pode ser visto, as stop words não mais se encontram no corpo texto, tendo em vista que, por exemplo, o ```as``` anteriormente presente não se verifica no texto abaixo, estando no lugar dele já a palavra retrowave. 

````python
# Normalizando o texto com stemming 

stemming = PorterStemmer()

print(f"Antes de normalizar com o stemming:\n\n{corpus}\n\nDepois de normalizar com stemming:\n")

for word in tokenized_corpus_without_stopwords:
    normalized_corpus = stemming.stem(word)
    print(normalized_corpus, end=" ")
````

Analisando ambos os corpus do texto, pode-se verificar que a normalização com stemming está implementada, tendo em vista que, por exemplo, o termo ```genre``` presente no primeiro corpus não se apresenta dessa forma no segundo, encontrando-se assim: ```genr```. 

````python
# Normalizando com lemmatizaion 

lammatization = WordNetLemmatizer()

print(f"Antes de normalizar com o stemming:\n\n{corpus}\n\nDepois de normalizar com lemmatization:\n")

for word in tokenized_corpus_without_stopwords:
    print(lammatization.lemmatize(word), end=" ")
````

Como pode-se perceber, o lemmatization não coloca todos os termos segundo o seu radical, tal como o stemming, mantendo por exemplo o termo `genre` no segundo corpus tal como ele se apresenta no primeiro. No entanto, ele realiza a sua transformação, tal como no termo `elements`, que se torna `element`, ao passo de sua aplicação.
